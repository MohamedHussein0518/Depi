# File Handling
### file modes
1. “r” --> **read mode**: reads the content of the file.
2. “a” --> **append mode**: creates the file if not exists, without overwriting its content
3. “w” --> **write mode**: creates the file if not exists, and overwrite its content.
4. “x” --> **excute mode**: creates the file only, and if it exists it will raise an error.

**Open a file using 'with' clause**

```python
import os
import random


def thanos_project(folder_path, num_files):
    """
    thanos project: create a folder, fill it with files, delete a random
    half of them, then report the result.

    :param folder_path: path of the folder to create and populate
    :type folder_path: str
    :param num_files: number of files to create inside the folder
    :type num_files: int
    :return: the list of file names remaining in the folder after the deletion
    :rtype: list
    """
    # 1- open the whole folder
    if not os.path.exists(folder_path):
        os.makedirs(folder_path)

    # 2- create number of files inside it
    for i in range(num_files):
        file_path = os.path.join(folder_path, f"file_{i}.txt")
        open(file_path, "a").close()

    # 3- check the len of files in that folder
    files = os.listdir(folder_path)
    print(f"Number of files before deletion: {len(files)}")

    # 4- delete random half of this folder
    files_to_delete = random.sample(files, len(files) // 2)
    for file_name in files_to_delete:
        os.remove(os.path.join(folder_path, file_name))

    # 5- check the result
    remaining_files = os.listdir(folder_path)
    print(f"Number of files after deletion: {len(remaining_files)}")
    return remaining_files


thanos_project("tmp_directory_thanos", 20)
```

## Task - per day: Basic Calculator

**Problem:** Create a simple calculator program that can perform basic arithmetic operations: addition, subtraction, multiplication, and division.

**Requirements:**
- Prompt the user to choose an operation: Addition, Subtraction, Multiplication, Division, or Exit.
- Validate that the entered numbers are numeric; re-prompt on invalid input.
- Check for division by zero.
- After each calculation, ask the user whether to continue or exit.
- Print the result in a friendly format, e.g. "The result of adding 5 and 3 is 8."
- On exit, print a thank you message.

```python
def display_menu():
    """
    Display the calculator's operation menu to the user.

    :return: None
    :rtype: None
    """
    print("\nPlease choose an operation:")
    print("1. Addition")
    print("2. Subtraction")
    print("3. Multiplication")
    print("4. Division")
    print("5. Exit")


def get_valid_number(prompt):
    """
    Prompt the user for a number and keep asking until a valid number is entered.

    :param prompt: message displayed to the user when asking for input
    :type prompt: str
    :return: the valid number entered by the user
    :rtype: float
    """
    while True:
        value = input(prompt)
        try:
            return float(value)
        except ValueError:
            print("Invalid input. Please enter a numeric value.")


def add(num1, num2):
    """
    Add two numbers together.

    :param num1: first number
    :type num1: float
    :param num2: second number
    :type num2: float
    :return: the sum of num1 and num2
    :rtype: float
    """
    return num1 + num2


def subtract(num1, num2):
    """
    Subtract the second number from the first.

    :param num1: first number
    :type num1: float
    :param num2: second number
    :type num2: float
    :return: the difference between num1 and num2
    :rtype: float
    """
    return num1 - num2


def multiply(num1, num2):
    """
    Multiply two numbers together.

    :param num1: first number
    :type num1: float
    :param num2: second number
    :type num2: float
    :return: the product of num1 and num2
    :rtype: float
    """
    return num1 * num2


def divide(num1, num2):
    """
    Divide the first number by the second.

    :param num1: numerator
    :type num1: float
    :param num2: denominator
    :type num2: float
    :raise ZeroDivisionError: if num2 is zero
    :return: the quotient of num1 divided by num2
    :rtype: float
    """
    if num2 == 0:
        raise ZeroDivisionError("Cannot divide by zero.")
    return num1 / num2
```

```python
def calculator():
    """
    Run the basic calculator program: repeatedly show the menu, read two
    numbers, perform the chosen operation, then ask the user whether to
    continue or exit.

    :return: None
    :rtype: None
    """
    operations = {
        "1": ("adding", add),
        "2": ("subtracting", subtract),
        "3": ("multiplying", multiply),
        "4": ("dividing", divide),
    }

    while True:
        display_menu()
        choice = input("Enter your choice (1-5): ").strip()

        if choice == "5":
            print("Thank you for using the calculator. Goodbye!")
            break

        if choice not in operations:
            print("Invalid choice. Please select a number between 1 and 5.")
            continue

        action_name, operation = operations[choice]
        num1 = get_valid_number("Enter the first number: ")
        num2 = get_valid_number("Enter the second number: ")

        try:
            result = operation(num1, num2)
            print(f"The result of {action_name} {num1} and {num2} is {result}.")
        except ZeroDivisionError as e:
            print(f"Error: {e}")

        again = input("Do you want to perform another calculation? (yes/no): ").strip().lower()
        if again != "yes":
            print("Thank you for using the calculator. Goodbye!")
            break


if __name__ == "__main__":
    calculator()
```
