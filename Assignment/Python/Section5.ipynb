```python

def sum(a, b):
    return a + b
def subtract(a, b):
    return a - b 
def multiply(a, b):
    return a * b
def divide(a, b):
    if b == 0:
        raise ValueError("Cannot divide by zero.")
    return a / b
```

```python
class StudentInfo:
    def __init__(self, name:str = None, age:int = 0, student_id:str = None):
        self.name = name
        self.age = age
        self.student_id = student_id

    def display_info(self):
        return f"Name: {self.name}, Age: {self.age}, Student ID: {self.student_id}"
    
```

```python
s1 = StudentInfo()
s1.name = input("Enter your name: ")
s1.age = int(input("Enter your age: "))
s1.student_id = input("Enter your student ID: ")
s1.display_info()
```

```python
s2 = StudentInfo("Alice", 20 ) 
s2.display_info()
```

```python
print(s1)
print(s2)
```

```python
from abc import ABC, abstractmethod
class Animal(ABC):
    @abstractmethod
    def move(self):
        pass
class Bird(Animal):
    def move(self):
        return "move from bird"
class Cat(Animal):
    def move(self):
        return "move from cat"
```

```python
a = Bird()
a.move()
```

```python
class Employee:
    def __init__(self, name:str, salary:int):
        self.name = name
        self.salary = salary
    def set_salary(self, obj ,new_salary:int):
        obj.salary = new_salary
```
