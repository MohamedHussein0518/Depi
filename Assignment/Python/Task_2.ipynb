# Task 1

```python
email = "Amit_ml@gmail.edu"
"""Parses and validates an email address to extract user and domain details.

    Checks if the provided email string follows a basic valid format (contains 
    exactly one '@' and a domain dot). It then extracts the username, identifies 
    the domain name, categorizes the domain type based on its extension (.com, 
    .edu, etc.), and prints the parsed information.

    Args:
        email (str): The email address string to be parsed and validated.

    Returns:
        None: This function prints the parsed email components directly to the console.
    """
if email.count("@") == 1 and "." in email.split("@")[1]:

    username = email.split("@")[0]
    domain_part = email.split("@")[1]  
    domain_components = domain_part.split(".")  
    domain = ".".join(domain_components[:-1]) 

    if email.endswith(".com"):
        domain_type = "Commercial Domain"
    elif email.endswith(".edu"):
        domain_type = "Educational Domain"
    else:
        domain_type = "Other Domain"

    print(f"Username: {username}")
    print(f"Domain: {domain}")
    print(f"Domain Type: {domain_type}")

else:
    print("Invalid email")
```

```python
encoded = "&&&**$gnirtS PLIO!!@1234"
"""Parses, reverses, and modifies specific characters within an encoded payload.

    Slices a predefined substring from the raw encoded string, splits it into 
    separate word components, reverses the target scrambled word, and performs 
    character replacements on the trailing string component.

    Args:
        encoded_str (str): The raw input string containing noise characters 
            and encoded text.

    Returns:
        None: Prints the sliced substring, reversed word, and fully 
            transformed message directly to the console.
    """
text = encoded[6:18]
Sp_text = text.split(" ")
re_text = Sp_text[0][::-1]

rep_text = Sp_text[1].replace("I", "E").replace("O", "U")
print(text)
print(re_text)
print(f"{re_text} {rep_text}")
```
