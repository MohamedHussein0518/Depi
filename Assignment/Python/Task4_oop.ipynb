## Task 1: Email Parser

```python
class Task1:

    def __init__(self, email: str):

        self.email = email

    def parse_email(self):

        if self.email.count("@") == 1 and "." in self.email.split("@")[1]:
            username = self.email.split("@")[0]
            domain_part = self.email.split("@")[1]
            domain_components = domain_part.split(".")
            domain = ".".join(domain_components[:-1])

            if self.email.endswith(".com"):
                domain_type = "Commercial Domain"
            elif self.email.endswith(".edu"):
                domain_type = "Educational Domain"
            else:
                domain_type = "Other Domain"

            print(f"Username: {username}")
            print(f"Domain: {domain}")
            print(f"Domain Type: {domain_type}")
        else:
            print("Invalid email")


task1 = Task1("Amit_ml@gmail.edu")
task1.parse_email()
```

## Task 5: Multiplication Table

## Task 6: Twin Primes

```python
class Task6:


    def __init__(self, limit: int):

        self.limit = limit

    def is_prime(self, n):

        if n < 2:
            return False
        for i in range(2, int(n ** 0.5) + 1):
            if n % i == 0:
                return False
        return True

    def find_twin_primes(self):

        print("Twin Prime:")
        for n in range(3, self.limit, 2):
            if self.is_prime(n) and self.is_prime(n + 2):
                print(f"{n} and {n + 2}")


task6 = Task6(1000)
task6.find_twin_primes()
```

## Task 7: Prime Factors

```python
class Task7:


    def __init__(self, number: int):

        self.number = number

    def prime_factors(self):

        num = self.number
        factors = []
        divisor = 2
        while num > 1:
            while num % divisor == 0:
                factors.append(divisor)
                num //= divisor
            divisor += 1
        return factors


task7 = Task7(56)
print(task7.prime_factors())
```

## Task 8: Decimal to Binary

```python
class Task8:


    def __init__(self, number: int):

        self.number = number

    def to_binary(self):

        num = self.number
        if num == 0:
            return "0"
        binary = ""
        while num > 0:
            binary = str(num % 2) + binary
            num //= 2
        return binary


task8 = Task8(11)
print(task8.to_binary())
```

## Task 9: Perfect Numbers

```python
class Task9:


    def __init__(self, start: int, end: int):

        self.start = start
        self.end = end

    def find_perfect_numbers(self):

        for num in range(self.start, self.end + 1):
            divisors_sum = sum(i for i in range(1, num) if num % i == 0)
            if divisors_sum == num:
                print(num)


task9 = Task9(0, 100)
task9.find_perfect_numbers()
```
