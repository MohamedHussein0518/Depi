```python
num1 = float(input("Enter the first number: "))
num2 = float(input("Enter the second number: "))

add = num1 + num2
sub = num1 - num2
multi = num1 * num2

print("\n--- Mathematical Results ---")
print(f"Addition: {num1} + {num2} = {add}")
print(f"Subtraction: {num1} - {num2} = {sub}")
print(f"Multiplication: {num1} * {num2} = {multi}")

if num2 != 0:
    div = num1 / num2
    print(f"Division: {num1} / {num2} = {div}")
else:
    print("Division: Cannot divide by zero!")
```

```python
score = float(input("Enter the student's degree: "))

if 0 <= score <= 100:
    if score >= 90:
        grade = 'A'
    elif score >= 80:
        grade = 'B'
    elif score >= 70:
        grade = 'C'
    elif score >= 60:
        grade = 'D'
    else:
        grade = 'F'
    
    print(f"The student's grade is: {grade}")
else:
    print("Invalid degree! Please enter a value between 0 and 100.")
```
